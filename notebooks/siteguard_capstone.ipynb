## 1. Custom data training — real Construction-PPE dataset


```python
from ultralytics import YOLO
model = YOLO("yolo26n.pt")
train_results = model.train(data="construction-ppe.yaml", epochs=50, imgsz=640, batch=16, patience=12, seed=42, plots=True, project="runs/siteguard", name="ppe_train", exist_ok=True)
```

## 2. Real validation and concrete metrics


```python
best = YOLO("runs/siteguard/ppe_train/weights/best.pt")
metrics = best.val(data="construction-ppe.yaml", imgsz=640, conf=0.35, iou=0.50, plots=True, project="runs/siteguard", name="evaluation", exist_ok=True)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", getattr(metrics.box, "mp", None))
print("Recall:", getattr(metrics.box, "mr", None))
```
